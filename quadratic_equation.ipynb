```python
from langgraph.graph import START, END, StateGraph
from typing import TypedDict, Literal
```

```python
class QuadState(TypedDict):
    a: int
    b: int
    c: int

    equation: str
    discriminant: float
    answer: str
```

```python
def create_equation(state: QuadState):
    equation =  f"{state["a"]}x2 + {state["b"]}x + {state["c"]}"
    return {'equation': equation}
```

```python
def find_discriminant(state: QuadState):
    discriminant = state["b"]**2 - (4*state["a"]*state["c"])
    return {'discriminant': discriminant}

def real_roots(state: QuadState):
    root1 = (-state["b"] + state["discriminant"] ** 2)/(2*state["a"])
    root2 = (-state["b"] - state["discriminant"] ** 2)/(2*state["a"])

    answer = f'Roots are{root1} and {root2}'
    return { 'answer' : answer }

def repeated_roots(state: QuadState):
    root1 = (-state["b"] ) / (2*state["a"])
    answer = f'Only Root is {root1} '
    return { 'answer' : answer }

def no_real_roots(state: QuadState):
    return { 'answer' : 'No real roots' }

def check_condition(state: QuadState) -> Literal["real_roots", "repeated_roots", "no_real_roots"]:
    if state["discriminant"] > 0:
        return "real_roots"
    elif state["discriminant"] == 0:
        return "repeated_roots"
    else:
        return "no_real_roots"

   

```

```python
graph  = StateGraph(QuadState)

graph.add_node("create_equation", create_equation)
graph.add_node("find_discriminant", find_discriminant)
graph.add_node("real_roots", real_roots)
graph.add_node("repeated_roots", real_roots)
graph.add_node("no_real_roots", real_roots)

graph.add_edge(START,'create_equation')
graph.add_edge('create_equation', 'find_discriminant')
graph.add_conditional_edges('find_discriminant',check_condition )
graph.add_edge('real_roots', END)
graph.add_edge('repeated_roots', END)
graph.add_edge('no_real_roots', END)

workflow = graph.compile()
```

```python
initial_state = {
    'a': 4,
    'b': -5,
    'c': -4

}
workflow.invoke(initial_state)
```
